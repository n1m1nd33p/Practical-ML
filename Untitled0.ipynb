```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

df = pd.read_csv("/content/Advertising Budget and Sales.csv")

df.head()
```

```python
X_simple = df[['TV Ad Budget ($)']]
y_simple = df['Sales ($)']

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_simple, y_simple, test_size=0.2, random_state=42
)

model_simple = LinearRegression()
model_simple.fit(X_train_s, y_train_s)

y_pred_s = model_simple.predict(X_test_s)

print(f"Coefficient (Slope): {model_simple.coef_[0]:.2f}")
print(f"Intercept: {model_simple.intercept_:.2f}")
print(f"Mean Squared Error (MSE): {mean_squared_error(y_test_s, y_pred_s):.2f}")
print(f"R-squared Score: {r2_score(y_test_s, y_pred_s):.2f}")
```

```python
plt.figure(figsize=(10, 4))

plt.scatter(
    X_test_s,
    y_test_s,
    color='lightgray',
    alpha=0.5,
    label='Actual Data'
)

plt.plot(
    X_test_s,
    y_pred_s,
    color='crimson',
    linewidth=2,
    label='Regression Line'
)

plt.title('Simple Linear Regression: Predicting Sales from TV Ad Budget')
plt.xlabel('TV Ad Budget ($)')
plt.ylabel('Sales ($)')
plt.legend()
plt.show()
```

```python
X_multi = df[['TV Ad Budget ($)', 'Radio Ad Budget ($)', 'Newspaper Ad Budget ($)']]
y_multi = df['Sales ($)']

X_train_m, X_test_m, y_train_m, y_test_m = train_test_split(
    X_multi, y_multi, test_size=0.2, random_state=42
)

model_multi = LinearRegression()
model_multi.fit(X_train_m, y_train_m)

y_pred_m = model_multi.predict(X_test_m)

print("Coefficients:")
print(f" - TV Impact: {model_multi.coef_[0]:.2f}")
print(f" - Radio Impact: {model_multi.coef_[1]:.2f}")
print(f" - Newspaper Impact: {model_multi.coef_[2]:.2f}")
print(f"Intercept: {model_multi.intercept_:.2f}")
print(f"Mean Squared Error (MSE): {mean_squared_error(y_test_m, y_pred_m):.2f}")
print(f"R-squared Score: {r2_score(y_test_m, y_pred_m):.2f}")
```
