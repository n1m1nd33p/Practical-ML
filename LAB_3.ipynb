```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/df_final_features.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'].dropna(), bins=30, kde=True, color='royalblue')
plt.title('Distribution of Athlete Ages', fontsize=14)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df['Height'], y=df['Weight'], color='crimson', alpha=0.4)
plt.title('Scatter Plot: Height vs Weight', fontsize=14)
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['Age'], color='mediumseagreen')
plt.title('Box Plot of Age (Outlier Inspection)', fontsize=14)
plt.xlabel('Age')
plt.show()

plt.figure(figsize=(10, 6))
sns.boxplot(data=df[['Age', 'Height', 'Weight']])
plt.title('Box Plot of Athlete Biometrics', fontsize=14)
plt.show()
```

```python
plt.figure(figsize=(10, 8))

cols_to_correlate = ['Age', 'Height', 'Weight', 'Medal_Binary']
corr_matrix = df[cols_to_correlate].corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap (Selected Features)', fontsize=14)
plt.show()
```

```python

pairplot_cols = ['Age', 'Height', 'Weight', 'Medal_Binary']

sns.pairplot(df[pairplot_cols].dropna(), diag_kind='kde', corner=True, plot_kws={'alpha':0.2})
plt.suptitle('Pair Plot of Biometrics and Medals', y=1.02, fontsize=14)
plt.show()
```
