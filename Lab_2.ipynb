```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/student_exam_performance.csv")
```

```python
df.info()
df.shape
```

```python
df = df.drop_duplicates()
print(f"Dataset shape after removing duplicates: {df.shape}")
```

```python
df.isna().sum()
```

```python
df.describe
```

```python
df['parent_education'] = df['parent_education'].fillna(df['parent_education'].mode()[0])
df['exam_score'] = df['exam_score'].fillna(df['exam_score'].mean())
```

```python
df.isna().sum()
```

### Handling Remaining Missing Values and Encoding Categorical Features

```python
from sklearn.preprocessing import LabelEncoder
for col in ['previous_gpa', 'attendance_percentage', 'time_management_score']:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mean())
for col in ['notes_quality', 'sleep_quality', 'device_availability']:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mode()[0])
print(df[['previous_gpa', 'attendance_percentage', 'notes_quality', 'sleep_quality', 'device_availability', 'time_management_score']].isnull().sum())
```

```python
categorical_cols = df.select_dtypes(include='object').columns.tolist()
if 'student_id' in categorical_cols:
    categorical_cols.remove('student_id')
print(f"Categorical columns to be encoded: {categorical_cols}")
label_encoders = {}
for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le

print("\nDataFrame after Label Encoding:")
display(df.head())
```
